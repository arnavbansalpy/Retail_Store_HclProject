```python
import pandas as pd


df = pd.read_csv("ecommerce_orders.csv")
print(df.head(), "\n")

completed = df[df["order_status"] == "Completed"]

print("Total orders:", len(df))
print("Total completed-order value:", completed["order_value"].sum(), "\n")

print("Sales by category:")
print(completed.groupby("category")["order_value"].sum().sort_values(ascending=False), "\n")

print("Sales by city:")
print(completed.groupby("city")["order_value"].sum().sort_values(ascending=False), "\n")

orders_per_customer = df.groupby("customer_id")["order_id"].count()
repeat_percent = (orders_per_customer > 1).mean() * 100

print(f"Customers who placed more than one order: {repeat_percent:.1f}%\n")

print("Average order value by segment:")
print(completed.groupby("customer_segment")["order_value"].mean(), "\n")

print("Order status breakdown (%):")
print((df["order_status"].value_counts(normalize=True) * 100).round(1), "\n")

correlation = df["delivery_days"].corr(df["rating"])
print(f"Correlation between delivery time and rating: {correlation:.2f}")
print("(a negative number means longer delivery = lower ratings)\n")

orders_per_customer_segment = df.groupby(["customer_id", "customer_segment"])["order_id"].count()
repeat_rate = orders_per_customer_segment.groupby("customer_segment").apply(lambda x: (x > 1).mean() * 100)

print("Repeat-purchase rate by segment (%):")
print(repeat_rate.round(1), "\n")

print("Takeaways:")
print("- Focus on the top-selling category and city shown above.")
print("- If the correlation in Step 7 is negative, faster delivery should raise ratings.")
print("- Target the lower-repeat-rate segment with offers to boost repeat orders.")
```
